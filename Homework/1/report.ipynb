<b>1.</b> Create a program that will read in all the images from a folder and use them to get the calibration 
parameters. Calibrate your camera using a minimum of three data sets. Compare and contrast the 
results. Do they make sense. One of the data sets should have at least 25 images. Calibrate the 
camera using 2, 5, 10, 15 and 25 images. Compare the results. What do they tell us?

```python
import os
import random
import numpy as np
import cv2
import matplotlib.pyplot as plt
from tqdm.auto import tqdm

from src.calibration.calibrate_camera import CameraCalibrator
from src.calibration.chessboard_detector import ChessboardDetector
from src.io.image_reader import ImageReader
from src.processing.image_filtering import ImageFiltering
from src.processing.image_thresholding import Thresholder
from src.processing.morphological_filters import MorphologicalFilter

CAMERA_NAME = "me566_2"
CALIBRATION_PATH = "images/calibration_set_me566_2"
FILTERED_PATH = "images/filtered_set_me566_2"
OUTPUT_DIR = "outputs"
CALIBRATION_DIR = os.path.join(
    OUTPUT_DIR, 
    "callibration_matrices", 
    CAMERA_NAME
)
CHESSBOARD_SIZE = (10, 7)
SQUARE_SIZE_MM = 23.5

os.makedirs(CALIBRATION_DIR, exist_ok=True)
os.makedirs(OUTPUT_DIR, exist_ok=True)
```

```python
# Filter calibration images if needed and select one accepted/rejected example in the same pass.
source_images = ImageReader(CALIBRATION_PATH).images
detector = ChessboardDetector(chessboard_size=CHESSBOARD_SIZE)
write_filtered = not os.path.exists(FILTERED_PATH) or not os.listdir(FILTERED_PATH)
valid_count = 0

if write_filtered:
    os.makedirs(FILTERED_PATH, exist_ok=True)

good_example = bad_example = None
for image in tqdm(source_images, desc="Inspecting calibration images"):
    found, corners = detector.detect_chessboard(image)
    if write_filtered and found:
        cv2.imwrite(os.path.join(FILTERED_PATH, f"image_{valid_count}.jpg"), image)
        valid_count += 1
    if found and good_example is None:
        good_example = image.copy()
        cv2.drawChessboardCorners(good_example, CHESSBOARD_SIZE, corners, found)
    elif not found and bad_example is None:
        bad_example = image.copy()
    if not write_filtered and good_example is not None and bad_example is not None:
        break

if write_filtered:
    print(f"Saved {valid_count} images with detected chessboards.")
if good_example is None or bad_example is None:
    raise RuntimeError("Could not find both a detected and rejected chessboard image.")

filtered_dataset = list(ImageReader(FILTERED_PATH).images)
print(f"Using {len(filtered_dataset)} usable calibration images.")

fig, axes = plt.subplots(1, 2, figsize=(12, 5), constrained_layout=True)
for axis, example, title in zip(
    axes,
    (good_example, bad_example),
    ("Accepted: all chessboard corners detected", "Rejected: incomplete or unreliable corner detection"),
):
    axis.imshow(cv2.cvtColor(example, cv2.COLOR_BGR2RGB))
    axis.set_title(title)
    axis.axis("off")
fig.savefig(os.path.join(OUTPUT_DIR, "calibration_good_vs_bad.jpg"), dpi=300, bbox_inches="tight")
plt.show()

del detector, source_images, write_filtered, valid_count, good_example, bad_example, image, found, corners, fig, axes, axis, example, title
```

```python
# Use a fixed seed so each run compares the same image subsets.
rng = random.Random(566)
sample_sizes = [2, 5, 10, 15, 25]
calibration_sets = {size: rng.sample(filtered_dataset, size) for size in sample_sizes}
calibration_sets["all"] = filtered_dataset

def run_calibration(label, images):
    calibrator = CameraCalibrator(
        images=images,
        chessboard_size=CHESSBOARD_SIZE,
        square_size=SQUARE_SIZE_MM,
    )
    values = calibrator.calibrate()
    reprojection = calibrator.reprojection_error()
    file_label = values["successful_images"] if label == "all" else label
    path = os.path.join(CALIBRATION_DIR, f"{CAMERA_NAME}_{file_label}_image_cal.json")
    calibrator.save_calibration(path)

    return {
        "set size": label,
        "usable images": values["successful_images"],
        "RMS error": values["rms_error"],
        "mean reprojection error": reprojection["mean"],
        "saved to": path,
    }

calibration_results = [
    run_calibration(label, images)
    for label, images in tqdm(
        calibration_sets.items(),
        desc="Running calibrations",
    )
]

for result in calibration_results:
    print(f"\nCalibration set: {result['set size']}")
    print(f"Usable images: {result['usable images']}")
    print(f"RMS error: {result['RMS error']:.4f} pixels")
    print(f"Mean reprojection error: {result['mean reprojection error']:.4f} pixels")
    print(f"Saved calibration: {result['saved to']}")

# A single random subset can be unusually favorable or unfavorable.
# Repeat every subset size to measure the effect of sampling variation.
TRIALS_PER_SIZE = 100


def run_calibration_trials(sample_size, trials=TRIALS_PER_SIZE, seed=566):
    rng = random.Random(seed + sample_size)
    trial_results = []
    attempts = 0

    progress = tqdm(total=trials, desc=f"Trials ({sample_size} images)", leave=False)
    while len(trial_results) < trials:
        attempts += 1
        images = rng.sample(filtered_dataset, sample_size)
        calibrator = CameraCalibrator(
            images=images,
            chessboard_size=CHESSBOARD_SIZE,
            square_size=SQUARE_SIZE_MM,
        )
        try:
            values = calibrator.calibrate()
        except RuntimeError:
            continue
        reprojection = calibrator.reprojection_error()
        trial_results.append(
            {
                "rms_error": values["rms_error"],
                "reprojection_error": reprojection["mean"],
                "usable_images": values["successful_images"],
            }
        )
        progress.update(1)

    progress.close()
    rms_errors = np.array([result["rms_error"] for result in trial_results])
    reprojection_errors = np.array(
        [result["reprojection_error"] for result in trial_results]
    )
    return {
        "set size": sample_size,
        "trials": trials,
        "sampling attempts": attempts,
        "rejected draws": attempts - trials,
        "RMS mean": float(rms_errors.mean()),
        "RMS std": float(rms_errors.std(ddof=1)),
        "reprojection mean": float(reprojection_errors.mean()),
        "reprojection std": float(reprojection_errors.std(ddof=1)),
        "trial results": trial_results,
    }


calibration_trial_summaries = [
    run_calibration_trials(sample_size)
    for sample_size in tqdm(sample_sizes, desc="Repeating calibration subsets")
]

print(f"\nMean ± standard deviation across {TRIALS_PER_SIZE} random subsets")
for summary in calibration_trial_summaries:
    print(
        f"{summary['set size']:>2} images: "
        f"RMS = {summary['RMS mean']:.4f} ± {summary['RMS std']:.4f} px, "
        f"reprojection = {summary['reprojection mean']:.4f} ± "
        f"{summary['reprojection std']:.4f} px"
    )

fig, axes = plt.subplots(1, 2, figsize=(11, 4), constrained_layout=True)
sizes = [summary["set size"] for summary in calibration_trial_summaries]
axes[0].boxplot(
    [[trial["rms_error"] for trial in summary["trial results"]] for summary in calibration_trial_summaries],
    tick_labels=sizes,
    showmeans=True,
)
axes[0].set(title="Calibration RMS across random subsets", xlabel="Images per subset", ylabel="RMS error (px)")
axes[1].boxplot(
    [[trial["reprojection_error"] for trial in summary["trial results"]] for summary in calibration_trial_summaries],
    tick_labels=sizes,
    showmeans=True,
)
axes[1].set(title="Reprojection error across random subsets", xlabel="Images per subset", ylabel="Mean reprojection error (px)")
for axis in axes:
    axis.grid(alpha=0.3)
fig.savefig(
    os.path.join(OUTPUT_DIR, "calibration_trial_summary.jpg"),
    dpi=300,
    bbox_inches="tight",
)
plt.show()

# Calibration data and helper functions are no longer needed below this point.
del (
    calibration_results, calibration_trial_summaries, calibration_sets, filtered_dataset,
    run_calibration, run_calibration_trials, rng, sample_sizes, TRIALS_PER_SIZE,
    result, summary, sizes, fig, axes, axis,
)
```

**Answer:** One hundred successful random trials were run for each non-full subset size, and the box-and-whisker plots show the full distribution rather than one favorable draw. The 2-image subsets have the lowest mean RMS (0.4424 px), but their large spread (0.4629 px) and high outliers make them unreliable. The 5-, 10-, 15-, and 25-image RMS means are 0.5759, 0.6223, 0.6579, and 0.6305 px, with overlapping boxes and outliers. Thus, more images do not automatically reduce the calibration's in-sample residual: subsets differ in pose coverage, blur, lighting, and corner-localization quality, and adding diverse but noisier views can raise residual while improving representativeness. A stronger comparison would evaluate every calibration on held-out images and compare the stability of the recovered intrinsic parameters.

```python
NOISY_IMAGES = {
    "Camera man": "images/camera_man_w_noise.jpg",
    "Salt & pepper": "images/SaltAndPepperNoise.jpg",
    "Gaussian": "images/GaussianNoise.jpg",
    "Uniform": "images/UniformNoise.jpg",
}
noisy_images = {name: ImageReader(path).images[0] for name, path in NOISY_IMAGES.items()}


def show_image_row(images, title):
    fig, axes = plt.subplots(1, len(images), figsize=(4 * len(images), 4))
    for axis, (label, image) in zip(axes, images.items()):
        axis.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
        axis.set_title(label)
        axis.axis("off")
    fig.suptitle(title, fontsize=16)
    plt.tight_layout()
    filename = title.lower().replace(" ", "_").replace("&", "and")
    fig.savefig(os.path.join(OUTPUT_DIR, f"{filename}.jpg"), dpi=300, bbox_inches="tight")
    plt.show()


show_image_row(noisy_images, "Original noisy images")

# The path mapping has served its purpose; the image arrays remain for filtering below.
del NOISY_IMAGES

FILTERS = {
    "Gaussian 3x3": {"filter_type": "gaussian", "kernel_size": (3, 3)},
    "Gaussian 5x5": {"filter_type": "gaussian", "kernel_size": (5, 5)},
    "Median 3x3": {"filter_type": "median", "kernel_size": (3, 3)},
    "Median 5x5": {"filter_type": "median", "kernel_size": (5, 5)},
    # rank range based on kernal size. for 3x3 it is 0-8 and 5x5 is 0-24
    "Rank 0 3x3": {"filter_type": "rank", "kernel_size": (3, 3), "rank": 0},
    "Rank 4 3x3": {"filter_type": "rank", "kernel_size": (3, 3), "rank": 4},
    "Rank 8 3x3": {"filter_type": "rank", "kernel_size": (3, 3), "rank": 8},
    "Rank 0 5x5": {"filter_type": "rank", "kernel_size": (5, 5), "rank": 0},
    "Rank 12 5x5": {"filter_type": "rank", "kernel_size": (5, 5), "rank": 12},
    "Rank 24 5x5": {"filter_type": "rank", "kernel_size": (5, 5), "rank": 24},
}

filtered_results = {}
for label, settings in tqdm(FILTERS.items(), desc="Applying filters"):
    filtered_results[label] = {
        name: ImageFiltering(**settings).apply(image)
        for name, image in noisy_images.items()
    }

for label, images in tqdm(filtered_results.items(), desc="Saving filter comparisons"):
    show_image_row(images, label)

# All filter comparisons have been displayed and saved.
del FILTERS, filtered_results, noisy_images, show_image_row, label, images
```

**Answers (2–3):** `ImageFiltering.apply` compares Gaussian blur, median blur, and rank filtering using 3×3 and 5×5 windows. Gaussian filtering is the best general choice for Gaussian or uniform noise; 5×5 suppresses more noise but blurs edges more than 3×3. Median filtering is best for salt-and-pepper noise and preserves edges well; 3×3 is preferred unless the impulses are severe. A middle rank (rank 4 for 3×3 or 12 for 5×5) behaves like a median filter, while minimum/maximum ranks are useful only for one-sided dark/bright impulses. Since no clean reference image is available, quality is judged visually by residual noise versus preserved edges and detail; with a reference, PSNR, MSE, or SSIM would provide an objective metric.


<b>4.</b> Use the thresholding tutorial provided on 
https://docs.opencv.org/4.x/d7/d4d/tutorial_py_thresholding.html. Perform both simple and 
adaptive thresholding. Compare both methods and discuss the tradeoffs between them. Make sure 
to include multiple images demonstrating the findings discussed in the writeup. 

- Coins1 
- Coins2 
- screws

```python
THRESHOLD_IMAGES = {
    "Coins 1": "images/coins1.jpg",
    "Coins 2": "images/coins2.jpg",
    "Screws": "images/screws.jpeg",
}

threshold_sources = {
    name: ImageReader(path).images[0]
    for name, path in THRESHOLD_IMAGES.items()
}


def compare_thresholds(name, image):
    simple = Thresholder(image, "simple").run_thresholding()
    adaptive = Thresholder(image, "adaptive").run_thresholding()

    fig, axes = plt.subplots(
        1, 3,
        figsize=(15, 5),
        constrained_layout=True
    )

    axes[0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    axes[0].set_title("Original")

    axes[1].imshow(simple, cmap="gray", vmin=0, vmax=255)
    axes[1].set_title("Simple Threshold")

    axes[2].imshow(adaptive, cmap="gray", vmin=0, vmax=255)
    axes[2].set_title("Adaptive Threshold")

    for axis in axes:
        axis.axis("off")

    fig.suptitle(name, fontsize=15)
    filename = name.lower().replace(" ", "_")
    fig.savefig(os.path.join(OUTPUT_DIR, f"thresholds_{filename}.jpg"), dpi=300, bbox_inches="tight")
    plt.show()

    return {
        "Original": image,
        "Simple": simple,
        "Adaptive": adaptive,
    }


threshold_results = {
    name: compare_thresholds(name, image)
    for name, image in tqdm(threshold_sources.items(), desc="Comparing thresholds")
}

# Keep threshold_results for morphology, but release the source images and helper now.
del THRESHOLD_IMAGES, threshold_sources, compare_thresholds
```

**Answer:** Simple thresholding uses one global cutoff (127), so it gives clean results when illumination and foreground/background contrast are consistent. Adaptive Gaussian thresholding uses a local 11×11 neighborhood with a constant of 2, making it more reliable for uneven lighting in the coins and screws images. Its tradeoff is that texture and noise can create extra foreground pixels; therefore, use simple thresholding for uniform scenes and adaptive thresholding when lighting varies spatially.


<b>5.</b> Use morphological operators to improve the segmentation of the images in problem 3. There are 
multiple operations with different masks and orders. Describe which set of operations works best 
and why. Make sure to include all images generated. See 
https://docs.opencv.org/4.x/d9/d61/tutorial_py_morphological_ops.html for help on how to use 
the operators.

```python
MORPHOLOGY = ["erosion", "dilation", "opening", "closing", "gradient", "tophat", "blackhat"]
morph_filters = {operation: MorphologicalFilter(operation, kernel_size=3) for operation in MORPHOLOGY}


def show_morphology(dataset_name, thresholds):
    inputs = {name: image for name, image in thresholds.items() if name != "Original"}
    fig, axes = plt.subplots(len(morph_filters), len(inputs), figsize=(8, 3 * len(morph_filters)))
    for row, (operation, morph_filter) in enumerate(
        tqdm(morph_filters.items(), desc=f"Morphology: {dataset_name}", leave=False)
    ):
        for col, (threshold_name, image) in enumerate(inputs.items()):
            axes[row, col].imshow(morph_filter.apply(image), cmap="gray")
            axes[row, col].set_title(f"{operation.title()} — {threshold_name}")
            axes[row, col].axis("off")
    fig.suptitle(dataset_name, fontsize=16)
    plt.tight_layout()
    filename = dataset_name.lower().replace(" ", "_")
    fig.savefig(os.path.join(OUTPUT_DIR, f"morphology_{filename}.jpg"), dpi=300, bbox_inches="tight")
    plt.show()


for name, thresholds in tqdm(threshold_results.items(), desc="Saving morphology comparisons"):
    show_morphology(name, thresholds)

# Morphology is the final use of these thresholded images and filters.
del MORPHOLOGY, morph_filters, threshold_results, show_morphology, name, thresholds
```

**Answer:** The morphology comparison uses 3×3 square kernels on the simple and adaptive binary segmentations. Opening is most useful for removing isolated foreground noise, while closing fills small holes and reconnects small gaps in objects. For the cleanest object masks, use opening followed by closing: this removes speckles before repairing small breaks. Erosion and dilation are useful for shrinking or expanding masks, and gradient, top-hat, and black-hat are best treated as edge or small-feature visualizations rather than final segmentations.
